# Gemini Editor Toolkit — Kaggle Runner
Set `REPO_URL` to your GitHub repo URL, attach `GEMINI_API_KEY` (and optional `APIFY_API_TOKEN`) if running AI/Apify steps, or let the Telegram Agent inject `JOB_CONFIG`.

```python
import os, sys, subprocess

REPO_URL = os.getenv('TOOLKIT_REPO_URL', 'https://github.com/iammidhun6771/agent-helper-kaggle.git')
REPO_DIR = '/kaggle/working/Gemini_editor'

if 'YOUR_USER' not in REPO_URL:
    if not os.path.exists(REPO_DIR):
        subprocess.run(['git', 'clone', '--depth', '1', REPO_URL, REPO_DIR], check=True)
    else:
        subprocess.run(['git', '-C', REPO_DIR, 'pull', '--ff-only'], check=True)

if os.path.exists(os.path.join(REPO_DIR, 'requirements.txt')):
    subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', '-r', os.path.join(REPO_DIR, 'requirements.txt')], check=True)
    if REPO_DIR not in sys.path:
        sys.path.insert(0, REPO_DIR)
    import kaggle_toolkit as kt
    print('✅ kaggle_toolkit ready:', [x for x in dir(kt) if not x.startswith('_')])
else:
    print('⚠️ Set REPO_URL to your GitHub repository URL first.')
```

```python
# Execute either injected JOB_CONFIG action or custom inline function call
if 'JOB_CONFIG' in globals() and isinstance(JOB_CONFIG, dict):
    for k, v in (JOB_CONFIG.get('ephemeral_secrets') or {}).items():
        os.environ[str(k)] = str(v)
    if JOB_CONFIG.get('action') and 'kt' in globals():
        out = kt.dispatch_job_config(JOB_CONFIG)
        print('RESULT:', out)
    elif JOB_CONFIG.get('code'):
        exec(JOB_CONFIG['code'], globals())
else:
    print('Interactive mode: call kt.download_media(...), kt.remove_watermark(...), kt.transcribe_audio(...), etc.')
```
